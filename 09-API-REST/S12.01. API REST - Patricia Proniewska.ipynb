  <span style="display:block;">1. Consulta la API pública JSONPlaceholder utilizando el método GET para obtener:</span>
  <span style="display:block;">• Lista de publicaciones (/posts)</span>
  <span style="display:block;">• Lista de usuarios (/users)</span>
  <span style="display:block;">• Lista de tareas (/todos)</span>
</div>

```python
# Imports y configuración

import requests
import pandas as pd

BASE_URL = "https://jsonplaceholder.typicode.com"

#Para que pandas muestre mejor los DataFrames

pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 120)
```

```python
# GET posts
response_posts = requests.get(BASE_URL + "/posts")
posts = response_posts.json()

# GET users
response_users = requests.get(BASE_URL + "/users")
users = response_users.json()

# GET todos
response_todos = requests.get(BASE_URL + "/todos")
todos = response_todos.json()
```

```python
df_posts = pd.DataFrame(posts)
df_users = pd.DataFrame(users)
df_todos = pd.DataFrame(todos)
```

He hecho una petición GET a la API JSONPlaceholder para obtener posts, usuarios y tareas.

  <span style="display:block;">4. Realiza una petición POST para crear una nueva publicación ficticia.</span>
  <span style="display:block;">Incluye el título, el cuerpo del mensaje y un userId.</span>
  <span style="display:block;">• Muestra la respuesta JSON.</span>
  <span style="display:block;">• Muestra el código de estado.</span>
</div>

```python
# Creo diccionario con el post:
new_post = {
    "title": "Peli mala",
    "body": "Esta pelicula esta fatal, lo mas aburrido que he visto en mi vida!",
    "userId": 1986
}
```

```python
# Envio a la API:
response_post = requests.post(BASE_URL + "/posts", json=new_post)
```

```python
print(response_post.json())
```

```python
print("Status code:", response_post.status_code)
```

He reralizado una petición POST a la API para crear una nueva publicación ficticia.
He enviado los datos en formato JSON y la API ha devuelto el objeto creado junto con un código de estado 201 (creado correctamente).

  <span style="display:block;">5. Realiza una petición PATCH para modificar parcialmente una publicación existente.</span>
  <span style="display:block;">• Muestra la respuesta JSON.</span>
  <span style="display:block;">• Muestra el código de estado.</span>
</div>

```python
# Modifico el titulo del post (ej: id 1)
update_post = {
    "title": "Título nuevo ahora con PATCH"
}

response_patch = requests.patch(BASE_URL + "/posts/1", json=update_post)

# Respuesta JSON
print(response_patch.json())

# Status code
print("Status code:", response_patch.status_code)
```

He realizado una petición PATCH para modificar parcialmente una publicación existente (su titulo).
La API devolvió el objeto actualizado y un código de estado 200 (correcto.).
