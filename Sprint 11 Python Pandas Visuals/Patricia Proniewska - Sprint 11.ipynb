```python
histograma = sns.histplot(data=df_products, x="price_$USD", bins=30, edgecolor="black",linewidth=0.6)

histograma.set_title("Distribution of product prices (USD)", fontsize=14, weight="semibold")
histograma.set_xlabel("price (USD)", fontsize=11)
histograma.set_ylabel("number or products", fontsize=11)

sns.despine(ax=histograma, left=True)

plt.tight_layout()
plt.show()
```

```python
df = df_transactions.merge(df_users, left_on="user_id", right_on="id")

# Orden de continentes
continent_order = ["Europe", "America"]

# Lista de paises en cada continente
countries_europe = df[df["continent"] == "Europe"]["country"].unique().tolist()
countries_america = df[df["continent"] == "America"]["country"].unique().tolist()

# Unimos respetando orden visual
hue_order = countries_europe + countries_america

barplot = sns.barplot(
    data=df,
    x="continent",
    y="amount",
    hue="country",
    order=continent_order,
    hue_order=hue_order,
    palette="muted",
    edgecolor="black",
    linewidth=0.6,
    errorbar=None)

barplot.set_title("Average Transaction Amount by Country across Continents", fontsize=14, weight="semibold")
barplot.set_xlabel("continent", fontsize=11)
barplot.set_ylabel("avg amount", fontsize=11)

plt.legend(title="country", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.xticks(rotation=45, ha="right")

sns.despine()
plt.tight_layout()
plt.show()

```

El gráfico muestra el importe medio de las transacciones por país dentro de cada continente. En general, los valores son bastante parecidos entre los países europeos, sin diferencias muy grandes. En América, tanto Estados Unidos como Canadá tienen también importes medios similares a los de Europa. Esto sugiere que el gasto medio por transacción es bastante un uniforme entre países y continentes

```python
ventas_producto = df_transactionproduct["product_id"].value_counts().reset_index()
ventas_producto.columns = ["product_id", "times_sold"]

ventas_producto.head(10)
```

```python
df_products_sales = df_products.merge(ventas_producto, left_on="id", right_on="product_id", how="left")
```

```python
sns.set_theme(style="ticks")

pairplot = sns.pairplot(df_products_sales[["price_$USD", "weight", "times_sold"]], diag_kind="hist", plot_kws={"alpha": 0.8, "s": 30}, diag_kws={"edgecolor": "black"})

pairplot.fig.suptitle("Price, Weight and Sales Relationships", fontsize=14, weight="semibold", y=1.05)

plt.show()
```

El pairplot muestra cómo se relacionan el precio, el peso y el número de veces que se vende cada producto. En general, los puntos están bastante dispersos, lo que sugiere que no hay una relación fuerte entre estas variables. El precio y el peso - los puntos aparecen muy dispersos y no se aprecia una tendencia clara.Además, el número de ventas se mantiene en valores parecidos sin importar demasiado el precio o el peso del producto.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="white", palette="muted")

jointplot = sns.jointplot(data=df_products_sales, x="price_$USD", y="times_sold", kind="reg")

jointplot.fig.suptitle("Relation between price and number of sales", fontsize=14, weight="semibold", y=1.02)

plt.show()
```

Este jointplot muestra la relación entre el precio y el número de ventas. Aunque la linea de tendencia sube un poco, los puntos están bastante dispersos, lo que indica que la relación es muy débil. Esto significa que el precio no parece influir claramente - cuántas veces se vende un producto. Los histogramas ayudan a ver cómo se reparten los precios y las ventas por separado.
